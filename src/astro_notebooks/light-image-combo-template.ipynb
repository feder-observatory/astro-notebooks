# Light image combination template

Click on a code cell (has grey background) then press Shift-Enter (at the same time) to run a code cell. That will add the controls (buttons, etc) you use to do the reduction one-by-one; then use them for reduction.

```python
from pathlib import Path

from reducer.image_browser import ImageBrowser

from ccdproc import ImageFileCollection

from astro_notebooks.image_selector import ImageSelect, SelectedCombiner

from reducer import __version__
print(__version__)
```

# Enter name of directory that contains your data in the cell below 

Also set `run_label` below. It becomes the base of the name of the combined output file and of the manifest that records which frames went into it, so use a **different label** each time you combine a different set of selected frames, or you will overwrite the earlier result.

```python
data_dir = 'shifted'
destination_dir = 'combined'

# Base of the name of the combined image and of the manifest that lists
# the frames it was made from. Use a different label for a different set
# of selected frames.
run_label = 'combined_light'

path = Path(".") / destination_dir
path.mkdir(exist_ok=True)
```

# Load your data set and review images to include

Uncheck a frame to leave it out of the combination below. The frames that are combined are the ones checked **at the moment you press the combine button**, so you can run every cell first and review afterwards. Your selection is saved next to the data, in `image_selection.json`, and is restored the next time you open this notebook. No image files are moved or deleted.

```python
isel = ImageSelect(directory=data_dir)
isel
```

## Combine light images

Pressing the button combines the frames that are checked above at that moment. When the combination finishes, a manifest listing the frames that went into it is written next to the result, as `<run_label>_manifest.json`, and its location is shown below the button. If no frame is checked, nothing is combined and a message says so.

```python
lcom = SelectedCombiner(image_select=isel,
                        run_label=run_label,
                        description="Combine light images",
                        toggle_type='button',
                        group_by='filter',
                        apply_to={'imagetyp': 'light'},
                        destination=destination_dir)
lcom.display()
```

```python
print(lcom)
```

## Look at the result: do the combined images look different?

```python
reduced_collection = ImageFileCollection(location=destination_dir, keywords='*')
```

```python
reduced_browser = ImageBrowser(reduced_collection, keys=['imagetyp', 'exposure'])
reduced_browser.display()
```
