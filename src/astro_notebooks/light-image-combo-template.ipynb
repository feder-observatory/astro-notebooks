# Light image combination template

Click on a code cell (has grey background) then press Shift-Enter (at the same time) to run a code cell. That will add the controls (buttons, etc) you use to do the reduction one-by-one; then use them for reduction.

```python
from pathlib import Path

import reducer.astro_gui as astro_gui
from reducer.image_browser import ImageBrowser

from ccdproc import ImageFileCollection

from astro_notebooks.image_selector import ImageSelect, write_selection_manifest

from reducer import __version__
print(__version__)
```

# Enter name of directory that contains your data in the cell below 

Also set `run_label` below. It becomes the base of the name of the combined output file and of the manifest that records which frames went into it, so use a **different label** each time you combine a different set of selected frames, or you will overwrite the earlier result.

```python
data_dir = 'shifted'
destination_dir = 'combined'

# Base of the name of the combined image and of the manifest that lists
# the frames it was made from. Use a different label for a different set
# of selected frames.
run_label = 'combined_light'

path = Path(".") / destination_dir
path.mkdir(exist_ok=True)
```

# Load your data set and review images to include

Uncheck a frame to leave it out of the combination below. Your selection is saved next to the data, in `image_selection.json`, and is restored the next time you open this notebook. No image files are moved or deleted.

Click a thumbnail to see that frame at full resolution in the viewer on the right, along with close-ups of the stars that were measured on it. Each thumbnail also lists the FWHM (the width of the stars, in pixels) and how bright those stars are compared with the rest of the night. Values shown in red flag a frame whose stars are broader, or fainter, than in most of the other frames; those are the frames worth looking at and, if they really are bad, unchecking.

```python
isel = ImageSelect(directory=data_dir)
isel
```

## Combine light images

```python
images = ImageFileCollection(location=data_dir, filenames=isel.selected_files)

lcom = astro_gui.Combiner(description="Combine light images",
                          toggle_type='button',
                          file_name_base=run_label,
                          group_by='filter',
                          image_source=images,
                          apply_to={'imagetyp': 'light'},
                          destination=destination_dir)
lcom.display()
```

```python
print(lcom)
```

```python
# Record which frames went into this combination, next to the result.
manifest_path = write_selection_manifest(isel, destination_dir, run_label)
print(f'Wrote {manifest_path}')
```
