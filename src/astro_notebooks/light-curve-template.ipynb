# Light curve viewer

Click on a code cell (has grey background) then press Shift-Enter (at the same time) to run a code cell. Run the cells from the top down. The last cell shows the plot with the controls to its right.

```python
from astro_notebooks.lightcurve_viewer import LightCurveViewer
```

# Pick your files in the cell below

Run the cell, then use the two file choosers it shows. For each one, click **Select**, find the file, click on it, and click **Select** again.

* **Photometry file** is the file of measurements that stellarphot wrote, for example `photometry-transformed.ecsv`.
* **Source list** is the list of stars, `source_locations.ecsv`. It is used to name the stars in the list of stars; leave it unset if you do not have one.

```python
from ipyfilechooser import FileChooser

photometry_chooser = FileChooser(title='Photometry file', filter_pattern=['*.ecsv', '*.csv', '*.fits'])
source_chooser = FileChooser(title='Source list (optional)', filter_pattern=['*.ecsv', '*.csv', '*.fits'])
display(photometry_chooser, source_chooser)
```

# Look at your light curves

Use the controls on the right of the plot:

* **Passband** picks the filter whose images are plotted.
* **Show** picks what is plotted. A note under it says which choices are missing and why. Magnitudes are plotted with brighter at the top.
* **X axis** picks time (BJD), hours since the first image, or airmass.
* **Stars** picks the stars to plot. Hold Ctrl (Cmd on a Mac) or Shift to pick several.
* **Offset** moves each star up by a bit more than the one before it, so that the light curves do not sit on top of each other.
* **Error bars** turns the error bars on and off.
* **Title** adds a title to the plot.

```python
if photometry_chooser.selected is None:
    raise ValueError('Pick a photometry file in the cell above, then run this cell again.')

viewer = LightCurveViewer(photometry_chooser.selected, source_chooser.selected)
viewer
```
