# Look at your light curves

Use the controls on the right of the plot:

* **Passband** picks the filter whose images are plotted.
* **Show** picks what is plotted. A note under it says which choices are missing and why. Magnitudes are plotted with brighter at the top.
* **X axis** picks time (BJD), hours since the first image, or airmass.
* **Stars** picks the stars to plot. Hold Ctrl (Cmd on a Mac) or Shift to pick several.
* **Offset** moves each star up by a bit more than the one before it, so that the light curves do not sit on top of each other.
* **Error bars** turns the error bars on and off.
* **Title** adds a title to the plot.

```python
viewer = LightCurveViewer(photometry_file, source_file)
viewer
```
